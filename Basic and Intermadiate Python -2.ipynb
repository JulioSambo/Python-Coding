3. Write a python function fact(n) to compute $n!$ and then evaluate fact(10)/fact(5)

Mathematically we know that:

$fact(n) = n!$ = 
\begin{cases}
1 \, \, \text{if}\, \, \,  n = 0\\
n\cdot (n-1)! \, \, \text{if}\, \, \,   n > 0 
\end{cases}

```python
def fact(n):
    if n <= 0:
        return 1
    return n * fact(n-1)

print(fact(10) / fact(5))
```

4. Given the recursion relation: $x_{n+1} = x_n + x_{n-1}$, with $x_0, x_1 = 0, 1$. Write a suitable python script to generate all the values up to $x_{30}$. And calculate the series sum $\sum \frac{1}{x_n^2}$ Also estimate the Golden Ratio. 

```python
# This is called Fibonacci sequence.

x_0, x_1 = 0, 1
Fibonacci_sequence = [x_0, x_1]    # empty list to store the Fibonacci sequence
for i in range(2, 30):
    x_2 = x_0 + x_1
    x_0, x_1 = x_1, x_2  # update x_0 and x_1 for the next iteration
    Fibonacci_sequence.append(x_2)  # add the current Fibonacci number to the list
print('The first 30 numbers in the Fibonacci sequence are:\n', Fibonacci_sequence)


# Sum 
Squared_Fibonacci = [n**2 for n in Fibonacci_sequence]
Sum = 1 / sum(Squared_Fibonacci)
print(f"The sum is {Sum} ")

# Estimating the Golden Ratio using Fibonacci sequence
x_0, x_1 = 0, 1
Golden_list = []
for i in range(2, 30):
    x_2 = x_0 + x_1
    G_ratio = x_2 / x_1  # we calculate the ratio of the current Fibonacci number to the previous one       
    x_0, x_1 = x_1, x_2
    Golden_list.append(G_ratio)
print('The estimated values of the Golden Ratio using Fibonacci sequence are:\n', Golden_list)
```

8. Write a function that combines two lists by alternatingly taking elements: 
[a, b, c, d, e, f], [1, 2, 3, 4, 5, 6] $\rightarrow$ [a, 1, b, 2, c, 3, d, 4, e, 5, f, 6]   

```python
def Two_lists(list_1, list_2):
    if len(list_1) != len(list_2):
        return "The lists must have the same length."
    Mixed_list = []
    List = zip(list_1, list_2)  # list of pair tuples
    for paired_tuple in List:
        for i in paired_tuple:
            Mixed_list.append(i) 
    return Mixed_list


list_1 = ['a', 'b', 'c', 'd', 'e', 'f']
list_2 = [1, 2, 3, 4, 5, 6]
result = Two_lists(list_1, list_2)
print('The mixed list is:\n', result)
```

11. Create a set of 100 uniforme random numbers between 1 and 2. Find out the mean, variance, standard deviation, sem (standard error of the mean) for the set of the values. 
- Write codes and use appropriate function from numpy

```python
import math 
from random import uniform 

Numbers = [uniform(1, 2) for _ in range(100)]

S = 0
for number in Numbers:
    S+= number 
Mean = S/len(Numbers) 
print(f"The mean is: {Mean}")



# Variance 
Sum_squares = 0
for number in Numbers:
    Sum_squares += (number - Mean)**2
Variance = Sum_squares / len(Numbers)
print('The variance is', Variance)

# Standard Deviation 
Standard_deviation = math.sqrt(Variance)
print('The standard deviation is:', Standard_deviation)


# standard error of the mean
Sem = Standard_deviation / math.sqrt(len(Numbers))
print('The standard error of the mean is:', Sem)
```

14. Consider a vector $r = (1, \, -1, \, 2)$ in cartesian coordinates system. The vector is rotated by $\theta = 65^\circ$ with the following rotational operator:

$$ A = 
\begin{pmatrix}
\cos \theta & -\sin \theta & 0 \\
\sin \theta & \cos \theta & 0 \\
0          & 0            & 1
\end{pmatrix}$$

```python
import numpy as np 

theta = 65 * 180/np.pi         # define the given angle 

R = np.array([[np.cos(theta), -np.sin(theta), 0],
              [np.sin(theta), np.cos(theta), 0],
              [0, 0, 1]])                              # Rotational matrix 


r = np.array([1, -1, 2])                               # Rotational vector

# Rotate r
r_rotated = R @ r    # or r_rotated = np.dot(R, r)

print(f"The rotated vector is {r_rotated}")
```
